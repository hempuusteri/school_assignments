```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.model_selection import cross_val_score
```

```python
housing = pd.read_csv('housing5.csv')

housing = housing.drop("ocean_proximity", axis=1)
```

```python
X_housing =  housing.drop("median_house_value", axis=1) # drop labels for training set
y_housing = housing["median_house_value"] 
```

```python
X_housing_train, X_housing_test, y_housing_train, y_housing_test = train_test_split(X_housing, y_housing, test_size=0.2, random_state=42)
imputer = SimpleImputer(strategy='mean')
X_housing_train = imputer.fit_transform(X_housing_train)
X_housing_test = imputer.transform(X_housing_test)
```

```python
scaler = StandardScaler()
X_housing_train = scaler.fit_transform(X_housing_train)
X_housing_test = scaler.transform(X_housing_test)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
```

```python
# create a model for LinearRegression and train it
lin_reg =  LinearRegression()
lin_reg.fit(X_housing_train, y_housing_train)

# create a model for DecisionTreeRegressor and train
tree_reg = DecisionTreeRegressor(random_state=42)
tree_reg.fit(X_housing_train, y_housing_train)

# create a model for RandomForest and train
forest_reg = RandomForestRegressor(random_state=42)
forest_reg.fit(X_housing_train, y_housing_train)
```

```python
housing_predictions_lin = lin_reg.predict(X_housing_train)
print("Predictions LinearRegression:",housing_predictions_lin)
```

```python
housing_predictions_tree = tree_reg.predict(X_housing_train)
print("Predictions DecisionTreeRegressor:",housing_predictions_tree)
```

```python
housing_predictions_forest = forest_reg.predict(X_housing_train)
print("Predictions, Forest",housing_predictions_forest)
```

**4) Evaluate three models by calculating the Root Mean Squared Error (RMSE) on the training set and print their RMSE.** (2 points)

```python
from sklearn.metrics import mean_squared_error
lin_rmse = mean_squared_error(y_housing_train, housing_predictions_lin, squared=False)
tree_rmse = mean_squared_error(y_housing_train, housing_predictions_tree, squared=False)
forest_rmse = mean_squared_error(y_housing_train, housing_predictions_forest, squared=False)
lin_rmse
```

**5) Use k-fold cross-validation with linear regression, decision tree (regression), and random forest model**

```python
lin_scores = cross_val_score(lin_reg, X_housing_train, y_housing_train, scoring="neg_mean_squared_error", cv=10)
lin_rmse_scores = np.sqrt(-lin_scores)
lin_rmse_scores
```

```python
tree_scores = cross_val_score(tree_reg, X_housing_train, y_housing_train, scoring="neg_mean_squared_error", cv=10)
tree_rmse_scores = np.sqrt(-tree_scores)
tree_rmse_scores
```

```python
forest_scores = cross_val_score(forest_reg, X_housing_train, y_housing_train, scoring="neg_mean_squared_error", cv=10)
forest_rmse_scores = np.sqrt(-forest_scores)
forest_rmse_scores
```
