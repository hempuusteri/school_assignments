# Basics of Quantitative Graphs with Seaborn

```python
# Take libraries in use
import pandas as pd
import numpy as np
import matplotlib as mp
import matplotlib.pyplot as plt 
import seaborn as sns 
```

## Load dataset 

### Read iris (flowers) dataset  

```python
# Read iris data using Seaborn 'load_dataset'
df = sns.load_dataset('iris')
df.shape # present structure: number of rows and columns
```

# Graphs with Seaborn

## Histogram

```python
sns.histplot?
```

### Histogram where quantitative data is separeted by species  


```python
# parameter kde = True draws curve in histograms
# parameter hue separates data using qualitative data
sns.histplot(data = df, x= 'sepal_length',bins=15, hue = 'species', kde = True) 
plt.xlabel('Sepal length (cm)', fontsize=14)
plt.ylabel('Freq', fontsize=12)   
plt.title("Histogram for iris species", fontsize=18)
plt.show()
```

```python
# parameter multiple = 'dodge' draws onwn bar per qualitative data
sns.histplot(data = df, x= 'petal_length',bins=15, hue = 'species', kde = True, multiple = 'dodge') 
plt.xlabel('Petal length (cm)', fontsize=14)
plt.ylabel('Freq', fontsize=12)   
plt.title("Histogram for iris species", fontsize=18)
plt.show()
```

## Histogram for sepal width

```python
sns.histplot(data = df, x= 'petal_width',bins=12, hue = 'species', kde = True, multiple = 'dodge', binwidth=0.1) 
plt.xlabel('Petal width')
plt.ylabel('Freq', rotation = 0)   
plt.title("Histogram for iris species", fontsize=18)
plt.show()
```

## Box plot

```python
sns.boxplot?
```

```python

sns.boxplot(data = df, x= 'species', y = 'petal_length') 
plt.ylabel('Petal length (cm)', fontsize=12)
# plt.ylabel('', fontsize=12)   
plt.title("Boxplot for iris species", fontsize=18)
plt.show()
```

### Task 1: Create similar figures for Sepal length and width 

```python
sns.boxplot(data = df, x= 'species', y = 'sepal_length') 
plt.ylabel('sepal_length', fontsize=12)   
plt.title("Boxplot for iris species sepal length", fontsize=18)
plt.show()
sns.boxplot(data = df, x= 'species', y = 'sepal_width') 
plt.ylabel('sepal_width', fontsize=12)   
plt.title("Boxplot for iris species sepal width", fontsize=18)
plt.show()
sns.boxplot(data = df, x= 'sepal_length', y = 'sepal_width') 
plt.ylabel('sepal_width', fontsize=12)
plt.xlabel('sepal_length', fontsize=12)
plt.title("Boxplot for iris species sepal length and width", fontsize=18)
plt.show()
```

## Scatter Diagram

```python
sns.scatterplot?
```

### Scatter diagram with scatterplot

```python
# Two quantitative variables, one in x and another in y
sns.scatterplot(data = df, x= 'petal_length', y= 'sepal_length', hue = 'species') 
plt.xlabel('Petal length (cm)', fontsize=12)
plt.ylabel('Sepal length (cm)', fontsize=12)   
plt.title("Scatter diagram for iris species", fontsize=18)
plt.show()
```

### Scatter diagram with relplot


```python
sns.relplot?
```

```python
# With parameter col and row there is possible to get own graphs per qualitative variable 
# Parameter col_wrap restrict the number of graphs per row
sns.relplot(data = df, x= 'sepal_width', y= 'sepal_length', col = 'species', col_wrap = 2) 
```

### Seaborn Pairplot


```python
sns.pairplot?
```

```python
sns.pairplot(data = df) 
```

```python
# pairplot for selected variables
sns.pairplot(data = df, vars=['sepal_length', 'sepal_width', 'petal_length'])
plt.show()
```

## Pairplot and present petal_length and petal_width separated by species 

```python
sns.pairplot(data = df, vars =['petal_length', 'petal_width'], hue='species', corner=True)
plt.show()
```

```python
df = sns.load_dataset('tips')
print(df.head())
```

```python
#Histogram for total_bill
plt.hist(df['total_bill'], bins=20, color='darkorchid', edgecolor='mistyrose')
plt.xlabel('total bill')
plt.ylabel('freq')
plt.title('Histogram for total bill')
plt.show()
```

```python
#boxplots for total_bill per day
sns.boxplot(y = df['total_bill'], x = df['day'])
plt.xlabel('total bill')
plt.ylabel('day', rotation=0)
plt.title('Boxplot total bill per day')
plt.show()
```

```python
#bar graph for smoker per sex
sns.countplot(data = df, x = 'smoker', hue = 'sex')
plt.xlabel('smokers')
plt.title('Bargraph for smoker per sex')
plt.show()
```

```python
#scatter diagram for total_bill and tip
plt.scatter(df.total_bill,df.tip, color='plum')
plt.xlabel('total bill')
plt.ylabel('tips')
plt.title('Scatter diagram for total bills and tips')
plt.show()
```

```python
#scatter diagram for total_bill and tip per sex including linear line
sns.lmplot(data=df, x='total_bill', y='tip', hue='sex')
plt.xlabel('total bill')
plt.ylabel('tips')
plt.title('Scatter diagram for total bills and tips per sex')
plt.show()
```
