```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
#from scipy.stats import zscore
import warnings
warnings.filterwarnings('ignore')
```

```python
part0 = pd.read_csv('fraudTrain_part0.csv')
part1 = pd.read_csv('fraudTrain_part1.csv')
part2 = pd.read_csv('fraudTrain_part2.csv')
df = pd.concat([part0,part1,part2],axis=0)

pd.set_option('display.expand_frame_repr', False)
print(df.head(2))
```

```python
def feature_eng(df):
    df2 = df.drop('Unnamed: 0',axis=1)
    df2['time'] = pd.to_datetime(df2['trans_date_trans_time'])
    df2 = df2.drop('trans_date_trans_time',axis=1)
    df2['weekday'] = df2['time'].dt.day_name()
    df2['hour'] = df2['time'].dt.hour
    return df2


df2 = feature_eng(df)
```

```python
plt.pie(df2['is_fraud'].value_counts(), labels=["legit", "fraud"], autopct="%0.0f%%")
plt.show()
print(df2['is_fraud'].value_counts())

print(df2.groupby('is_fraud')['amt'].describe())
```

```python
cols = ['job', 'state', 'gender']
k = 1
plt.figure(figsize=(30, 10))

for i in cols:
    categ = df2.loc[df2['is_fraud'] == 1, i].value_counts().sort_values(ascending=False).reset_index().head(10)
    #print(categ)
    plt.subplot(1, 3, k)
    plt.bar(categ.iloc[:, 0], categ.iloc[:,1])
    plt.title(f'Frauds per {i} Categories')
    plt.xticks(rotation=45)
    k+= 1
plt.show()

cols = ['category', 'weekday', 'hour']
k = 1

plt.figure(figsize=(30, 10))

for i in cols:
    categ = df2.loc[df2['is_fraud'] == 1, i].value_counts().sort_values(ascending=False).reset_index().head(10)
    #print(categ)
    plt.subplot(1, 3, k)
    plt.bar(categ.iloc[:, 0], categ.iloc[:,1])
    plt.title(f'Frauds per {i}')
    plt.xticks(rotation=45)
    k+= 1
plt.show()
```

```python
plt.figure(figsize=(10,6))
corrmat = df2[['amt','hour','city_pop']].corr()
top_corr_features = corrmat.index
heat_map = sns.heatmap(corrmat, annot=True, cmap="viridis")
```

```python
cols=['amt','gender','state','category','city_pop','job','is_fraud','weekday','hour']
fraud_data_df = df2[cols]
print(fraud_data_df)
```

```python
def one_hot_encoded_cat(data):
    cat_cols = df2[cols].select_dtypes(include='object').columns
    for i in cat_cols:
        df1 = pd.get_dummies(data[i],prefix=i, drop_first=True)
        data.drop(i, axis=1, inplace=True)
        data = pd.concat([data, df1], axis=1)

    return data
```

```python
fraud_df = one_hot_encoded_cat(fraud_data_df)
print(fraud_df)
```
