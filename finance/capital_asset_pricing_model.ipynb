```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
pd.set_option('display.expand_frame_repr', False)
```

```python
tickers = pd.read_csv('sp500.csv')
stocks = tickers.iloc[:,1].to_list() + ["^GSPC"]#add sp500 index to list
```

```python
stocks_data = yf.download(stocks, start = '2024-01-01', end = '2025-12-31',auto_adjust=False)['Adj Close']
print(stocks_data)
```

```python
stocks_data.iloc[:,0:5].plot(figsize=(10, 6))
plt.legend(loc='upper left', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
stocks_data = np.log(stocks_data/stocks_data.shift())
print(stocks_data.head(10))
```

```python
corr = stocks_data.drop("^GSPC",axis=1).corr()#corr without sp500 index
corr.style.background_gradient(cmap='Blues')
```

```python
cov = stocks_data.cov()
print(cov)
```

```python
variance = stocks_data['^GSPC'].var()
print(variance)
```

```python
beta = cov.loc['^GSPC']/variance
print(beta)
```

```python
irx = yf.Ticker("^IRX").history(period="1d")["Close"] # annualized, percent
rf_annual = (irx / 100.0).rename("rf_annual")
risk_free_return = rf_annual.values
print(risk_free_return)
```

```python
market_return = 0.081
expected_return = risk_free_return + beta*(market_return - risk_free_return)
expected_return = expected_return.reset_index()
expected_return.rename(columns={"^GSPC": "ret"}, inplace=True)
print(expected_return)
```

```python
top = expected_return[expected_return["ret"] > 0.081]
top = top.sort_values(by="ret",ascending=False)
print(top)
```
