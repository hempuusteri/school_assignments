```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (classification_report,confusion_matrix, f1_score)
```

```python
df = pd.read_csv('fraudTrain_50.csv')
df2 = feature_eng(df)
cols=['amt','gender','state','category','city_pop','job','is_fraud','weekday','hour']
fraud_data_df = df2[cols]
fraud_df = one_hot_encoded_cat(fraud_data_df)
```

```python
X = fraud_df.drop('is_fraud',axis=1)
y = fraud_df['is_fraud']
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0,train_size=0.8)
X_train
```

```python
def reporter(y_test,preds):
    conf_mat = confusion_matrix(y_true=y_test,y_pred=preds)
    print('Confusion matrix:\n', conf_mat)
    print('--' * 25)
    print('Classification report:\n',classification_report(y_test, preds))
    return conf_mat
```

```python
param_log = {'C': np.logspace(-4, 4, 4), 'penalty': ['l1', 'l2']}
log_grid = GridSearchCV(LogisticRegression(),param_grid=param_log, n_jobs=-1)
log_grid.fit(X_train, y_train)
prediction_log = log_grid.predict(X_test)
```

```python
conf_mat_log = reporter(y_test,prediction_log)
```

```python
param_dt = {'max_depth': [3, 5, 10],'min_samples_split': [2, 4, 6],'criterion': ['gini', 'entropy']}
dt_grid = GridSearchCV(DecisionTreeClassifier(),param_grid=param_dt, n_jobs=-1)
dt_grid.fit(X_train, y_train)
prediction_dt = dt_grid.predict(X_test)
conf_mat_dt = reporter(y_test,prediction_dt)
```
