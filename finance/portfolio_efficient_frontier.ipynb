```python
def find_efficient_frontier(res):
    g = np.arange(res.volatility.min(),res.volatility.max(),(res.volatility.max()-res.volatility.min())/6)
    eff = 0*g
    tol = 0.005
    for i in range(0,len(g)):
        idx = np.logical_and(res.volatility < g[i]+tol,res.volatility > g[i]-tol)
        r = res.loc[idx,'returns'].max()
        eff[i] = r

    return g,eff
```

```python
g,eff = find_efficient_frontier(res)
plt.figure(figsize=(10, 7))
plt.scatter(res.volatility, res.returns, c=res.sharpe, cmap='YlGnBu', marker='o')
plt.xlabel('Volatility (Standard Deviation)')
plt.ylabel('Expected Return')
plt.colorbar(label='Sharpe Ratio')
plt.plot(g,eff,color='red',linewidth=3)
plt.grid(True)
plt.show()
```
