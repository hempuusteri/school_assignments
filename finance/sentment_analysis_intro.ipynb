```python
import nltk

from nltk.sentiment import SentimentIntensityAnalyzer
nltk.download('vader_lexicon')
```

```python
sid = SentimentIntensityAnalyzer()
text = "I hate this product! It's totally crap."
sentiment_score = sid.polarity_scores(text)
sentiment_score
```
