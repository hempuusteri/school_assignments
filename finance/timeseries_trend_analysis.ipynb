```python
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf
from datetime import datetime
```

```python
x = np.arange(0,6,0.1)
y = np.sin(x) + 0.05*np.random.randn(*x.shape)
plt.plot(x,y)
plt.show()
```

```python
def mov_avg(y,window_size):

    # Returns (centered) moving average of y using Pandas
    moving_averages = pd.Series(y).rolling(window_size,center=True).mean().tolist()
    y_mov = np.array(moving_averages).reshape(y.shape)

    return y_mov

plt.plot(x,mov_avg(y,9))
plt.show()
```

```python
end_date = datetime(year=2025, month=1, day=1)
start_date = datetime(year=2000, month=1, day=1)

df = yf.download(
tickers="AAPL",
start=start_date,
end=end_date,
interval="1d",
group_by="ticker",
auto_adjust=True,
progress=False

)
df.reset_index(inplace = True)
```

```python
plt.plot(df['Date'],df["AAPL"]["Close"])
plt.plot(df['Date'],mov_avg(df["AAPL"]["Close"],60))
plt.show()
```

```python
plt.plot(df['Date'],mov_avg(df["AAPL"]["Close"],180))
plt.plot(df['Date'],mov_avg(df["AAPL"]["Close"],360))
plt.show()
```

```python
def find_peaks(y):
    return np.where(np.diff(1*(np.diff(y)>0))==-1)
```

```python
def find_valleys(y):
    return np.where(np.diff(1*(np.diff(y)>0))==1)
```

```python
y = df['AAPL']['Close'].rolling(window=180).mean()

peaks_pos = find_peaks(y)
valleys_pos = find_valleys(y)
```

```python
print(np.max(np.diff(peaks_pos)))
print(np.max(np.diff(valleys_pos)))
```

```python
df = pd.read_csv('AirPassengers.csv')
ax = plt.plot(df.Month,df['#Passengers'])
ticks = plt.xticks()[0]
plt.xticks(ticks[::20])
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller

result = adfuller(df['#Passengers'])
print('ADF Statistic: %f' % result[0])
print('p-value: %f' % result[1])
print('Critical Values:')
for key, value in result[4].items():
    print('\t%s: %.3f' % (key, value))
```

```python
import yfinance as yf

hist = yf.Ticker("GOOGL").history(period="1y")
hist['Daily_Return'] = hist['Close'].pct_change()
plt.plot(hist['Daily_Return'])
plt.plot(hist['Daily_Return'].rolling(window=21).mean())
plt.plot(hist['Daily_Return'].rolling(window=21).var())
plt.legend(['GOOGL daily return','rolling mean','rolling variance'])
plt.show()
```
