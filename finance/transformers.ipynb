```python
import pandas as pd
from tqdm import tqdm
tqdm.pandas()

df3 = pd.read_csv('Reviews4.csv')
df3['Text'] = df3['Text'].str[:1000]
df3
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification, pipeline

tokenizer = AutoTokenizer.from_pretrained("nlptown/bert-base-multilingual-uncased-sentiment")

model = AutoModelForSequenceClassification.from_pretrained("nlptown/bert-base-multilingual-uncased-sentiment")
nlp = pipeline("sentiment-analysis", model=model, tokenizer=tokenizer)
```

```python
def analyzer(text):
    try:
        return nlp(text)[0].get('label')
    except:
        return"too many tokens in text"
```

```python
df3['Rating'] = df3['Text'].progress_apply(analyzer) 
```

```python
df4 = df3.sort_values('Rating')
print(df4.head())
```

```python
for lev in ['1 star','5 stars']:
    print(lev)
    idx = df3.loc[df3['Rating'] == lev].index
    for i in range(0,5):
        print(df3.iloc[idx[i]].Text)
```
